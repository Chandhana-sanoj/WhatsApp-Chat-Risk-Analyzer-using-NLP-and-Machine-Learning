```python
df = pd.read_csv(
    "/content/drive/MyDrive/Internship project/preprocessed_whatsapp_dataset.csv"
)

chat_inventory = pd.read_csv(
    "/content/drive/MyDrive/Internship project/chat_inventory.csv"
)
```

```python
chat_inventory = pd.read_excel(
    "/content/drive/MyDrive/Internship project/Book2.xlsx"
)

print(chat_inventory.shape)

chat_inventory.head()
```

```python
ground_truth = df.merge(
    chat_inventory,
    on=["Person", "Chat_Name"],
    how="left"
)

ground_truth.head()
```

```python
print(
    "Missing Chat Types:",
    ground_truth["Chat_Type"].isna().sum()
)
```

```python
ground_truth["Risk_Label"] = ""
ground_truth["Risk_Category"] = ""
ground_truth["Confidence"] = ""
ground_truth["Label_Source"] = ""
ground_truth["Review_Status"] = "Pending"
```

```python
ground_truth.loc[
    ground_truth["Chat_Type"]=="Normal",
    "Risk_Label"
] = "Normal"

ground_truth.loc[
    ground_truth["Chat_Type"]=="Normal",
    "Confidence"
] = "High"

ground_truth.loc[
    ground_truth["Chat_Type"]=="Normal",
    "Label_Source"
] = "Chat Inventory"

ground_truth.loc[
    ground_truth["Chat_Type"]=="Normal",
    "Review_Status"
] = "Reviewed"
```

```python
annotation_df = ground_truth[
    ground_truth["Chat_Type"]=="Suspicious"
].copy()

annotation_df.reset_index(
    drop=True,
    inplace=True
)

annotation_df.insert(
    0,
    "Message_ID",
    range(1,len(annotation_df)+1)
)
```

```python
annotation_df = annotation_df[
[
"Message_ID",
"Person",
"Chat_Name",
"Timestamp",
"Sender",
"Message",
"Risk_Label",
"Risk_Category",
"Confidence",
"Label_Source",
"Review_Status"
]
]
```

```python
annotation_df.to_excel(
    "/content/drive/MyDrive/Internship project/04_ground_truth_dataset.xlsx",
    index=False
)

print("Ground Truth Dataset Created Successfully!")
```
