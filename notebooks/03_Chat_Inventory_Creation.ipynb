```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(
    "/content/drive/MyDrive/Internship project/preprocessed_whatsapp_dataset.csv"
)
```

```python
chat_inventory = (
    df[["Person","Chat_Name"]]
    .drop_duplicates()
    .sort_values(["Person","Chat_Name"])
    .reset_index(drop=True)
)

chat_inventory["Chat_Type"] = ""
```

```python
print("Total Chats:", len(chat_inventory))

chat_inventory.groupby("Person").size()
```

```python
chat_inventory.to_excel(
    "/content/drive/MyDrive/Internship project/03_chat_inventory.xlsx",
    index=False
)

print("Saved successfully!")
```
