```python
df["Timestamp_Original"] = df["Timestamp"]
```

```python
df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    format="%d/%m/%y, %I:%M %p",
    errors="coerce"
)
```

```python
df = df.dropna(subset=["Timestamp"])
```

```python
print("Invalid timestamps:", df["Timestamp"].isna().sum())
```

```python
# Validate timestamps first
temp = pd.to_datetime(
    df["Timestamp"],
    format="%d/%m/%y, %I:%M %p",
    errors="coerce"
)

invalid = df[temp.isna()]

print(f"Invalid timestamps: {len(invalid)}")

if len(invalid) > 0:
    display(invalid[["Person", "Chat_Name", "Timestamp"]])
    raise ValueError("Fix the invalid timestamps before continuing.")

# Safe conversion
df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    format="%d/%m/%y, %I:%M %p"
)
```

```python
df["Date"] = df["Timestamp"].dt.date

df["Time"] = df["Timestamp"].dt.time

df["Year"] = df["Timestamp"].dt.year

df["Month"] = df["Timestamp"].dt.month

df["Month_Name"] = df["Timestamp"].dt.month_name()

df["Day"] = df["Timestamp"].dt.day

df["Weekday"] = df["Timestamp"].dt.day_name()

df["Hour"] = df["Timestamp"].dt.hour

df["Minute"] = df["Timestamp"].dt.minute
```

```python
df["Message"] = (
    df["Message"]
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
df["Is_Media"] = (
    df["Message"]
    .str.contains("media omitted", case=False, na=False)
)
```

```python
df["Is_Location"] = (
    df["Message"]
    .str.contains("location", case=False, na=False)
)
```

```python
df["Is_Deleted"] = (
    df["Message"]
    .str.contains("deleted", case=False, na=False)
)
```

```python
df["Is_Call"] = (
    df["Message"]
    .str.contains("call", case=False, na=False)
)
```

```python
df["Is_URL"] = (
    df["Message"]
    .str.contains(r"http|www", case=False, regex=True)
)
```

```python
df["Character_Count"] = df["Message"].str.len()

df["Word_Count"] = (
    df["Message"]
    .str.split()
    .str.len()
)
```

```python
df["Is_Weekend"] = df["Weekday"].isin([
    "Saturday",
    "Sunday"
])
```

```python
df["Is_Night"] = (
    (df["Hour"] >= 22) |
    (df["Hour"] <= 5)
)
```
